```python
from huggingface_hub import snapshot_download

local_dir = "fleurs_en_us_raw"

snapshot_download(
    repo_id="google/fleurs",
    repo_type="dataset",
    local_dir=local_dir,
    local_dir_use_symlinks=False,
    allow_patterns=[
        "data/en_us/*",   
    ],
)
```

```python
ROOT_DIR = r"fleurs_en_us_raw" 
LANG = "en_us"

META_EXT = ".tsv"  
SPLITS = ["train", "dev", "test"]
AUDIO_DIR = os.path.join(ROOT_DIR, "data", LANG, "audio")
META_DIR = os.path.join(ROOT_DIR, "data", LANG)


def extract_audio(split: str):
    tar_name = f"{split}.tar.gz"       
    tar_path = os.path.join(AUDIO_DIR, tar_name)
    out_dir = os.path.join(AUDIO_DIR, split)

    if not os.path.exists(tar_path):
        raise FileNotFoundError(f"Не знайдено архів {tar_path}")

    os.makedirs(out_dir, exist_ok=True)
    print(f"[{split}] Розпаковую {tar_path} → {out_dir}")

    import tarfile
    with tarfile.open(tar_path, "r:gz") as tar:
        for member in tar.getmembers():
            if not member.isfile():
                continue
            fname = os.path.basename(member.name)
            fileobj = tar.extractfile(member)
            if fileobj is None:
                continue
            out_path = os.path.join(out_dir, fname)
            with open(out_path, "wb") as f:
                f.write(fileobj.read())

    print(f"[{split}] Готово.")


def build_manifest(split: str):
    import pandas as pd
    import csv

    meta_path = os.path.join(META_DIR, f"{split}{META_EXT}")  
    if not os.path.exists(meta_path):
        raise FileNotFoundError(f"Не знайдено мета-файл для {split}: {meta_path}")

    print(f"[{split}] Читаю метадані з {meta_path}")

    df = pd.read_csv(
        meta_path,
        sep="\t",
        header=None,
        names=[
            "id",
            "file_name",
            "raw_transcription",
            "transcription",
            "phonemes",
            "num_samples",
            "gender",
        ],
    )

    df["audio_path"] = df["file_name"].apply(
        lambda fn: os.path.join("data", LANG, "audio", split, fn)
    )
    df["split"] = split

    out_manifest = os.path.join(ROOT_DIR, f"fleurs_en_us_{split}_manifest.csv")
    df[
        [
            "id",
            "file_name",
            "raw_transcription",
            "transcription",
            "num_samples",
            "gender",
            "audio_path",
            "split",
        ]
    ].to_csv(out_manifest, index=False, quoting=csv.QUOTE_MINIMAL)

    print(f"[{split}] Маніфест збережено в {out_manifest}")
    print(df.head(3))


def main():
    for split in SPLITS:
        extract_audio(split)
        build_manifest(split)


if __name__ == "__main__":
    main()
```

**Model**

```python
import os
import torch
import librosa
import numpy as np
import pandas as pd

from torch.utils.data import Dataset

from transformers import (
    WhisperProcessor,
    WhisperForConditionalGeneration,
    Seq2SeqTrainingArguments,
    Seq2SeqTrainer,
)
from dataclasses import dataclass
from typing import Dict, List, Any
```

```python
MODEL_ID = "openai/whisper-tiny.en"  

processor = WhisperProcessor.from_pretrained(MODEL_ID, language="en", task="transcribe")
model = WhisperForConditionalGeneration.from_pretrained(MODEL_ID)

for p in model.model.encoder.parameters():
    p.requires_grad = False

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
```

```python
class WhisperFleursDataset(Dataset):
    def __init__(self, csv_path: str, processor: WhisperProcessor, audio_root: str = "."):
        self.df = pd.read_csv(csv_path)
        self.processor = processor
        self.audio_root = audio_root

    def __len__(self):
        return len(self.df)

    def _get_full_path(self, rel_path: str) -> str:
        if os.path.isabs(rel_path):
            return rel_path
        return os.path.join(self.audio_root, rel_path)

    def __getitem__(self, idx: int) -> Dict[str, Any]:
        row = self.df.iloc[idx]
        audio_rel = row["audio_path"]
        audio_path = self._get_full_path(audio_rel)


        if not os.path.exists(audio_path):
            raise FileNotFoundError(f"Audio not found: {audio_path}")

        text = row["transcription"]

        wav, sr = librosa.load(audio_path, sr=16000)
        input_features = self.processor.feature_extractor(
            wav, sampling_rate=sr
        ).input_features[0]
        labels = self.processor.tokenizer(text).input_ids

        return {"input_features": input_features, "labels": labels}
```

```python
AUDIO_ROOT = "fleurs_en_us_raw"  

train_dataset = WhisperFleursDataset(
    "fleurs_en_us_raw/fleurs_en_us_train_manifest.csv",
    processor=processor,
    audio_root=AUDIO_ROOT,
)

val_dataset = WhisperFleursDataset(
    "fleurs_en_us_raw/fleurs_en_us_dev_manifest.csv",
    processor=processor,
    audio_root=AUDIO_ROOT,
)
```

```python
@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: WhisperProcessor

    def __call__(self, features: List[Dict[str, Any]]) -> Dict[str, torch.Tensor]:
        input_features = [f["input_features"] for f in features]
        label_features = [f["labels"] for f in features]

        batch = {}
        batch["input_features"] = torch.tensor(input_features, dtype=torch.float32)

        labels_batch = self.processor.tokenizer.pad(
            {"input_ids": label_features},
            padding=True,
            return_tensors="pt",
        )
        labels = labels_batch["input_ids"]
        labels[labels_batch["attention_mask"] == 0] = -100

        batch["labels"] = labels
        return batch

data_collator = DataCollatorSpeechSeq2SeqWithPadding(processor)
```

```python
training_args = Seq2SeqTrainingArguments(
    output_dir="./whisper-tiny-fleurs-en",
    per_device_train_batch_size=8,        
    per_device_eval_batch_size=8,
    gradient_accumulation_steps=2,       
    learning_rate=1e-4,
    num_train_epochs=3,
    logging_steps=50,
    eval_strategy="epoch",
    save_strategy="epoch",
    fp16=torch.cuda.is_available(),
    predict_with_generate=True,
    generation_max_length=225,
    report_to="none",
)
```

```python
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    data_collator=data_collator,
    tokenizer=processor.tokenizer,
)

trainer.train()
```

```python
save_dir = "./whisper-tiny-fleurs-en-finetuned"

trainer.save_model(save_dir)      
processor.save_pretrained(save_dir)
```

```python
from transformers import WhisperForConditionalGeneration, WhisperProcessor

MODEL_DIR = "./whisper-tiny-fleurs-en-finetuned"

processor = WhisperProcessor.from_pretrained(MODEL_DIR)
model = WhisperForConditionalGeneration.from_pretrained(MODEL_DIR)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()
```

```python
def transcribe_file(path: str) -> str:
    audio, sr = librosa.load(path, sr=16000)

    input_features = processor.feature_extractor(
        audio, sampling_rate=sr
    ).input_features
    input_features = torch.tensor(input_features, dtype=torch.float32).to(device)

    with torch.no_grad():
        predicted_ids = model.generate(input_features)

    text = processor.tokenizer.batch_decode(
        predicted_ids, skip_special_tokens=True
    )[0]

    return text

print(transcribe_file(r"fleurs_en_us_raw\data\en_us\audio\test\4821372520807131.wav"))
```

```python
from jiwer import wer, cer

MODEL_DIR = "./whisper-tiny-fleurs-en-finetuned"

processor = WhisperProcessor.from_pretrained(MODEL_DIR)
model = WhisperForConditionalGeneration.from_pretrained(MODEL_DIR)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()
```

```python
class WhisperFleursDataset(Dataset):
    def __init__(self, csv_path, processor, audio_root="."):
        self.df = pd.read_csv(csv_path)
        self.processor = processor
        self.audio_root = audio_root

    def __len__(self):
        return len(self.df)

    def _get_full_path(self, rel):
        if os.path.isabs(rel):
            return rel
        return os.path.join(self.audio_root, rel)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        audio_path = self._get_full_path(row["audio_path"])
        text = row["transcription"]

        wav, sr = librosa.load(audio_path, sr=16000)

        return {
            "audio": wav,
            "sr": sr,
            "text": text,
            "audio_path": audio_path
        }
```

```python
AUDIO_ROOT = "fleurs_en_us_raw"

test_dataset = WhisperFleursDataset(
    "fleurs_en_us_raw/fleurs_en_us_test_manifest.csv",
    processor=processor,
    audio_root=AUDIO_ROOT,
)
```

```python
def transcribe_audio(wav, sr=16000):
    input_features = processor.feature_extractor(
        wav, sampling_rate=sr
    ).input_features
    input_features = torch.tensor(input_features, dtype=torch.float32).to(device)

    with torch.no_grad():
        pred_ids = model.generate(input_features)

    text = processor.tokenizer.batch_decode(pred_ids, skip_special_tokens=True)[0]
    return text.lower().strip()
```

```python
pred_texts = []
true_texts = []

for i in range(len(test_dataset)):
    item = test_dataset[i]
    wav, sr = item["audio"], item["sr"]
    true_text = item["text"].lower().strip()

    pred = transcribe_audio(wav, sr)

    pred_texts.append(pred)
    true_texts.append(true_text)

    if i % 20 == 0:
        print(f"{i}/{len(test_dataset)} processed")
```

```python
test_wer = wer(true_texts, pred_texts)
test_cer = cer(true_texts, pred_texts)

print("Test WER:", test_wer)
print("Test CER:", test_cer)
```
